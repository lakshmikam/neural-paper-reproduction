# Reproduce Lagaris et al.
Run in Kaggle or Jupyter with internet enabled. CPU is sufficient. This trains all cases, solves the FEM baselines and verifies checkpoints; it does not need credentials or external datasets. These cells have not been executed on Kaggle; the corresponding scripts were executed locally.

```python
import pathlib, subprocess, sys
project = pathlib.Path.cwd()
if not (project / 'train.py').exists():
    project = project / 'neural-paper-reproduction'
    if not project.exists():
        subprocess.run(['git', 'clone', 'https://github.com/lakshmikam/neural-paper-reproduction.git', str(project)], check=True)
print(project)
```

```python
# Install the tested package versions.
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', str(project / 'requirements.txt')], check=True)
```

```python
# This overwrites the cloned reference results with your own measurements.
for script in ['train.py', 'fem.py', 'verify_artifacts.py', 'report.py']:
    subprocess.run([sys.executable, script], cwd=project, check=True)
print((project / 'results' / 'REPORT.md').read_text())
```
